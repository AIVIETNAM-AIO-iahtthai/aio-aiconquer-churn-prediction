```python
import pandas as pd

# Đọc file dữ liệu đã được làm sạch bước đầu
df = pd.read_csv("/cleaned_data.csv")
```

```python
# 2 cột này đang là chữ, cần ép về số (float)
df['monthlycharges'] = pd.to_numeric(df['monthlycharges'], errors='coerce')
df['totalcharges'] = pd.to_numeric(df['totalcharges'], errors='coerce')
```

```python
# Khi ép kiểu, những dòng chứa chữ/lỗi gõ sai sẽ biến thành khoảng trống (NaN).
# Chúng ta sẽ lấp đầy các khoảng trống mới sinh ra này bằng giá trị Trung vị (Median).
df['monthlycharges'] = df['monthlycharges'].fillna(df['monthlycharges'].median())
df['totalcharges'] = df['totalcharges'].fillna(df['totalcharges'].median())
```

```python
# TỐI ƯU CHUỖI NGÀY THÁNG & THÊM CỘT SỐ NGÀY SỬ DỤNG
# =========================================================
# Chuyển cột signup_date từ chuỗi văn bản (String) sang kiểu Thời gian (Datetime) chuẩn của hệ thống
df['signup_date'] = pd.to_datetime(df['signup_date'])

# Tìm ngày đăng ký gần nhất để làm mốc "hiện tại"
current_date = df['signup_date'].max()

# Tạo cột mới tính số ngày khách hàng đã gắn bó
df['days_since_signup'] = (current_date - df['signup_date']).dt.days
```

```python
# KIỂM TRA LẠI TOÀN BỘ & LƯU FILE

print("--- TỔNG SỐ Ô TRỐNG HIỆN TẠI ---")
# Kết quả lý tưởng phải là 0
print(df.isna().sum().sum())

print("\n--- KIỂM TRA KIỂU DỮ LIỆU MỚI ---")
# Kiểm tra lại các cột vừa xử lý xem đã đúng định dạng chưa
print(df[['monthlycharges', 'totalcharges', 'signup_date', 'days_since_signup']].info())
```

```python
# Lưu thành một file mới để báo cáo hoàn thành Task
df.to_csv("formatted_data.csv", index=False)
print("\nlưu file formatted_data.csv")
```
